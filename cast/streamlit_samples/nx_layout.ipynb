```python
import networkx as nx
import matplotlib.pyplot as plt

# 创建一个适合演示的图
G = nx.sedgewick_maze_graph()  # 一个小型的迷宫图，结构清晰
# G = nx.barbell_graph(m1=5, m2=2)

layouts = {
    "Spring": nx.spring_layout,
    "Circular": nx.circular_layout,
    "Kamada-Kawai": nx.kamada_kawai_layout,
    "Shell": nx.shell_layout,
    "Random": nx.random_layout,
    "Spectral": nx.spectral_layout,
    "Spiral": nx.spiral_layout,
    "Planar": nx.planar_layout
}

plt.figure(figsize=(15, 8))

for i, (name, layout_func) in enumerate(layouts.items()):
    plt.subplot(2, 4, i + 1)
    try:
        if name == "Shell":
            # Shell 需要简单的 nlist 避免重叠太严重，这里仅作默认演示
            pos = layout_func(G)
        else:
            pos = layout_func(G)

        nx.draw(G, pos, node_size=100, node_color='#40a6d1', edge_color='#999999', with_labels=False)
        plt.title(name)
    except Exception as e:
        plt.title(f"{name} (Error: Not applicable)")

plt.tight_layout()
plt.show()
```
